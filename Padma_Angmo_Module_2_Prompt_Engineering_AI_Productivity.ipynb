## Day 6 — Prompt Engineering Fundamentals
A strong prompt may contain **Role + Context + Instruction + Constraints + Output Format**.

**Zero-shot prompting:** asks for a task without examples.  
**Few-shot prompting:** provides examples to demonstrate the expected pattern.

```python
zero_shot = "Classify as Positive, Negative, or Neutral: 'The workshop was useful and easy to follow.'"

few_shot = '''Classify each sentence as Positive, Negative, or Neutral.
Example: "The session was excellent." -> Positive
Example: "The app keeps crashing." -> Negative
Example: "The meeting is at 4 PM." -> Neutral
Now classify: "The workshop was useful and easy to follow."'''

print("ZERO-SHOT PROMPT:\n", zero_shot)
print("\nFEW-SHOT PROMPT:\n", few_shot)
```

```python
role = "You are a professional study assistant."
context = "The user is a second-year CSE student learning Operating Systems."
instruction = "Summarize process scheduling."
constraints = "Use simple language, maximum 6 bullet points, and include one example."
prompt = f"Role: {role}\nContext: {context}\nInstruction: {instruction}\nConstraints: {constraints}"
print(prompt)
```

## Day 7 — Structured Outputs & Common Tasks
Structured outputs such as JSON make information easier to process. Prompt engineering can also be used for summarization, classification, and generation.

```python
import json
expected = {"name":"Padma","branch":"CSE (AIML)","skill":"Python"}
json_prompt = '''Extract Name, Branch and Skill from a student profile.
Return ONLY valid JSON with keys: name, branch, skill.'''
print(json_prompt)
print("\nExpected format:")
print(json.dumps(expected, indent=2))
```

## Day 8 — AI-Assisted Coding & Debugging
AI can help explain code, identify likely bugs, suggest corrections, and create test cases. AI-generated code must still be reviewed and tested.

```python
buggy_code = '''def average(numbers):
    total = 0
    for n in numbers:
        total += n
    return total / len(numbers)

print(average([]))'''

debug_prompt = f'''Act as a Python debugging assistant.
Find the problem, explain it, provide a safe corrected version, and give two test cases.
CODE:
{buggy_code}'''
print(debug_prompt)

def average(numbers):
    if not numbers:
        return 0
    return sum(numbers)/len(numbers)

print("\nCorrected-code tests:")
print("Test 1:", average([10,20,30]))
print("Test 2:", average([]))
```

## Day 9 — Reusable Templates & Prompt Evaluation
Reusable templates improve consistency. Prompts can be evaluated using clarity, specificity, relevance, correctness, completeness, and format compliance.

```python
def build_prompt(role, context, instruction, constraints, output_format):
    return f'''ROLE: {role}
CONTEXT: {context}
INSTRUCTION: {instruction}
CONSTRAINTS: {constraints}
OUTPUT FORMAT: {output_format}'''

print(build_prompt(
 "Research assistant",
 "A beginner is learning Generative AI.",
 "Explain prompt engineering.",
 "Use simple words and avoid unnecessary jargon.",
 "5 bullet points"
))
```

```python
report = [
 {"prompt":"Explain AI","clarity":2,"specificity":1,"format_control":1},
 {"prompt":"Explain AI to a beginner in 5 bullets with one example","clarity":5,"specificity":5,"format_control":5}
]
for x in report:
    x["average_score"]=round((x["clarity"]+x["specificity"]+x["format_control"])/3,2)
print(json.dumps(report,indent=2))
```
